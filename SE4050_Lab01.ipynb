## Task 1: Python Libraries Exercises

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from mpl_toolkits.mplot3d import Axes3D

# Ensure plots show inside notebook
%matplotlib inline
```

### 1.1 Create a random 4x4 array from exponential distribution

```python
arr = np.random.exponential(scale=1.0, size=(4,4))
print("4x4 Exponential Array:\n", arr)
```

### 1.2 Random distributions comparison

```python
exp_data = np.random.exponential(scale=1.0, size=100000)
norm_data = np.random.normal(size=100000)
uni_data = np.random.rand(100000)

plt.hist(exp_data, bins=100, density=True, histtype="step", label="Exponential")
plt.hist(norm_data, bins=100, density=True, histtype="step", label="Normal")
plt.hist(uni_data, bins=100, density=True, histtype="step", label="Uniform")

plt.legend()
plt.title("Random Distributions")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

### 1.3 3D plot of Z = X² + Y²

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap="viridis")
plt.show()
```

### 1.4 Correlation heatmaps

```python
# Example dataset (replace with correct dataset if given)
df = sns.load_dataset("penguins").dropna()

corr_pearson = df.corr(method="pearson")
corr_spearman = df.corr(method="spearman")

plt.figure(figsize=(10,5))
sns.heatmap(corr_pearson, annot=True, cmap="coolwarm")
plt.title("Pearson Correlation")
plt.show()

plt.figure(figsize=(10,5))
sns.heatmap(corr_spearman, annot=True, cmap="coolwarm")
plt.title("Spearman Correlation")
plt.show()
```

## Task 2: Au Nanoparticle Dataset

```python
# Load your dataset (upload Au_nanoParticle_dataset.csv to Colab before running)
df = pd.read_csv("Au_nanoParticle_dataset.csv")
new_df = df[["N_total", "N_bulk", "N_surface", "R_avg"]]
new_df.head(20)
```

### 2.4 Advanced PairGrid (Hist + KDE)

```python
g = sns.PairGrid(new_df)
g.map_upper(sns.histplot)
g.map_diag(sns.histplot, kde=True)
g.map_lower(sns.kdeplot)
plt.show()
```
